Bu küçük örnekte bilerek bir etiket hatası oluşturup traceback'i okuyacağız. Gerçek veri veya erişim anahtarı kullanmayın.

```python
examples = [
    {"text": "Bu film çok iyiydi", "label": "positive"},
    {"text": "Bu film sıkıcıydı", "label": "negative"},
]
label_to_id = {"negative": 0, "positive": 1}
for example in examples:
    print(example, "→", label_to_id[example["label"]])
```

## Kasıtlı hata
Son satırdaki `KeyError` mesajını okuyun. Hangi girdinin beklendiğini ve hangi girdinin geldiğini karşılaştırın.

```python
def encode_label(example):
    return label_to_id[example["category"]]

encode_label(examples[0])  # Bilerek KeyError üretir
```

```python
def encode_label_fixed(example):
    return label_to_id[example["label"]]

assert [encode_label_fixed(item) for item in examples] == [1, 0]
print("Düzeltme doğrulandı")
```

## Sorun bildirimi alıştırması
**Başlık:** `encode_label` fonksiyonunda `KeyError: 'category'`

**Ortam:** Python sürümü ve çalıştırma ortamı

**Yeniden üretim:** Yukarıdaki veri tanımı ve hatalı fonksiyon

**Gerçek davranış:** `KeyError` oluşuyor.

**Beklenen davranış:** `label` alanından tamsayı etiket elde ediliyor.

**Düzeltme:** `category` yerine `label` anahtarı kullanılıyor.
